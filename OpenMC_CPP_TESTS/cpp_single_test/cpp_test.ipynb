```c++17
#include<iostream>

// using the unique_ptr
#include<memory>

// using the vector
#include<vector>

// code starts here
int main()
{
    // creating a unique pointer
    // std::unique_ptr<int> ptr(new int(10));

    // make_unique is used to create a unique pointer
    std::unique_ptr<int> ptr = std::make_unique<int>(10);
    
    // creating a vector of unique pointers
    std::vector<std::unique_ptr<int>> vec;
    
    // pushing the unique pointer into the vector
    vec.push_back(std::move(ptr));
    
    // printing the value of the unique pointer
    std::cout<<*vec[0]<<std::endl;
    
    return 0;
}
// code ends here
main();
```

```c++17
#include<iostream>

class abstract_a{
    public:
    virtual void print() = 0;
}

class a: public abstract_a{
    public:
    void print(){
        std::cout<<"Hello World"<<std::endl;
    }
}

int main(){
    a obj;
    obj.print();
    return 0;
}
```
